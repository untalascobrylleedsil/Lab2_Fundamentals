```python
# [CELL 1] CONFIGURATION
ENGINEER = "Brylle Untalasco"
STUDENT_ID = "TUPM-24-1234"

# Extract last digit to use as seed
SEED_NUM = int(STUDENT_ID[-1])

print(f"System Initialized for: {ENGINEER}")
print(f"Algorithmic Seed Value: {SEED_NUM}")
```

```python
# [CELL 2] FLOATING-POINT PRECISION
a = 0.1 + 0.2
b = 0.3

print("0.1 + 0.2 =", a)
print("0.3 =", b)
print("Are they equal?", a == b)
```

```python
# [CELL 3] REDACTION PROTOCOL
target_name = "UNTALASCO"  # your surname in uppercase
redacted_name = ""

for char in target_name:
    if char in "AEIOU":
        redacted_name += str(SEED_NUM)
    else:
        redacted_name += char

print("Original Name:", target_name)
print("Redacted Name:", redacted_name)
```

```python
# [CELL 4] VERIFICATION HASH
first_letter = ENGINEER[0]  # 'B'
ascii_value = ord(first_letter)  # Convert to ASCII
total_loops = len(target_name)  # Must run Cell 3 first

verification_hash = ascii_value * SEED_NUM * total_loops
print("Verification Hash:", verification_hash)
```

```python
# [CELL 5] PASSWORD LOOP
correct_password = "TUPM123"
attempts = 0
while True:
    user_input = input("Enter password: ")
    attempts += 1
    if user_input == correct_password:
        print("Access Granted!")
        break
    else:
        print("Incorrect password. Try again.")
print("Total Attempts:", attempts)
```
