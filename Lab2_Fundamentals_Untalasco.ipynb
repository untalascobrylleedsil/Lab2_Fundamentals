```python
# [CELL 1] CONFIGURATION

ENGINEER = "Brylle Untalasco"
STUDENT_ID = "TUPM-24-1234"

# Extract the last digit to act as Logic Seed
SEED_NUM = int(STUDENT_ID[-1])

print(f"System Initialized for: {ENGINEER}")
print(f"Algorithmic Seed Value: {SEED_NUM}")
```

```python
# [CELL 2] FLOATING-POINT PRECISION ANALYSIS

a = 0.1 + 0.2
b = 0.3

print(f"0.1 + 0.2 = {a}")
print(f"0.3 = {b}")
print(f"Are they equal? {a == b}")
```

```python
# [CELL 3] REDACTION PROTOCOL

original_name = "UNTALASCO"  # your surname in uppercase
redacted_name = ""

vowels = "AEIOU"

for char in original_name:
    if char in vowels:
        redacted_name += str(SEED_NUM)  # replace vowel with seed
    else:
        redacted_name += char  # keep consonant

print(f"Original Name: {original_name}")
print(f"Redacted Name: {redacted_name}")
```

```python
# [CELL 4] VERIFICATION HASH

first_letter = ENGINEER[0]  # first letter of your name
ascii_value = ord(first_letter)  # convert to ASCII

# Example: total loops = length of surname
total_loops = len(original_name)

verification_hash = ascii_value * SEED_NUM * total_loops

print(f"First Letter ASCII: {ascii_value}")
print(f"SEED_NUM: {SEED_NUM}")
print(f"Total Loops: {total_loops}")
print(f"Verification Hash: {verification_hash}")
```

```python
# [CELL 5] PASSWORD LOOP

correct_password = "TUPM1234"
attempts = 0

while True:
    user_input = input("Enter password: ")
    attempts += 1
    if user_input == correct_password:
        print(f"Password correct! Attempts: {attempts}")
        break
    else:
        print("Incorrect password, try again.")
```
