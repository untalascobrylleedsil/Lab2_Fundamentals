```python
# [APPENDIX 1] REDACTION PROTOCOL

# Setup
target_name = "UNTALASCO"   # Your surname in uppercase
SEED_NUM = 4                # Last digit of STUDENT_ID
redacted_name = ""
execution_log = []          # Initialize log list

# Redaction loop
for char in target_name:
    if char in "AEIOU":
        redacted_name += str(SEED_NUM)
        execution_log.append(f"Vowel {char} replaced with {SEED_NUM}")
    else:
        redacted_name += char
        execution_log.append(f"Consonant {char} left unchanged")

# Outputs
print("Original Surname:", target_name)
print("SEED_NUM Used:", SEED_NUM)
print("Execution Log:")
for log in execution_log:
    print(log)
print("Redacted Output:", redacted_name)
```

```python
# [APPENDIX 2] VERIFICATION HASH

first_letter = "B"  # First letter of ENGINEER
ascii_value = ord(first_letter)  # Convert to ASCII
total_loops = len(target_name)
execution_log = []

# Calculate hash with logging
verification_hash = 0
for i in range(total_loops):
    verification_hash += ascii_value * SEED_NUM
    execution_log.append(f"Loop {i+1}: hash += {ascii_value}*{SEED_NUM}")

# Outputs
print("First Letter ASCII:", ascii_value)
print("SEED_NUM:", SEED_NUM)
print("Total Loops:", total_loops)
print("Execution Log:")
for log in execution_log:
    print(log)
print("Verification Hash:", verification_hash)
```

```python
# [APPENDIX 3] PASSWORD LOOP

correct_password = "1234"
attempts = 0
execution_log = []

while True:
    user_input = input("Enter password: ")
    attempts += 1
    execution_log.append(f"Attempt {attempts}: entered '{user_input}'")
    if user_input == correct_password:
        execution_log.append("Password correct! Exiting loop.")
        break
    else:
        execution_log.append("Password incorrect. Try again.")

# Outputs
print("Correct Password Set:", correct_password)
print("Input Attempts:", attempts)
print("Execution Log:")
for log in execution_log:
    print(log)
print("Attempts Before Success:", attempts)
```
